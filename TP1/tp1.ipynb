```python
#QUESTION 1
x = np.array([4,6,2,5,8])

#all entries
print(x)

#all entries except first and last
Nn_first = x[1:]
Nn_last = x[:-1]
print(Nn_first, Nn_last)

#every second entry, starting with first and entries at reverse order
sliced_array = x[0::2]
back = x[::-1]
print(sliced_array, back)

#selection with indeces
chosen_ind = x[[0,2,4]]

#last element
last = x[-1]
print(chosen_ind, "\n", last)

#sequence wiht np arrange and np linspace
seq = np.arange(2, 12, 2)
lspace = np.linspace(0, 1, 11)
print(seq, lspace, "\n")


#Question 2
E = np.array([
    [2,4,1,5],
    [6,7,2,1],
    [3,5,8,9]
])

print(E, "\n")

#vectro x1 and matrix y
X1 = E[:, 1]
Y = E[:, [1,3]]
print(X1,"\n", Y)

#sum of each columns/rows
sum_c = E.sum(axis=0)
sum_r = E.sum(axis=1)
print('\n', sum_c, sum_r)

#shapes of E X1 and Y
sh_E = E.shape
sh_x1 = X1.shape
sh_Y = Y.shape
print('\n', sh_E, sh_x1, sh_Y)

#zero matrix
Z = np.zeros((3,4))
print("n",Z)


```
