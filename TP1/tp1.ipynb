```python
#Question 4
An = np.arange(2,21,2)
An_diag = np.diag(An)

#print(An_diag)

#Use np.zeros, np.eye and np.vstack to place one zero row above the order10 identity matrix.
Z = np.zeros((1,10))
#Identy matrix
Id = np.eye(10)
#print("\n",Id,"\n\n",Z, "\n\n")
B = np.vstack((Z,Id))
up_shape = B.shape
#print(B, "\n", up_shape)

#removing the last row
B_updt = B[:-1,:]
B_transpose = B_updt.T
print("\n", B_updt, "\n\n", B_transpose)
```

```python
#Combined matrix
final = An_diag + B_updt - B_transpose
print(final)
```
